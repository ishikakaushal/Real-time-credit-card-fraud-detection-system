## Part 3 — Class Imbalance

Credit card fraud detection is a highly imbalanced classification problem.

The majority of transactions are legitimate, while only a small proportion
are fraudulent.

In our dataset:

- Class 0 represents legitimate transactions.
- Class 1 represents fraudulent transactions.

Because the fraud class is significantly smaller than the legitimate class,
accuracy alone is not an appropriate metric for evaluating model performance.

Therefore, we will focus on precision, recall, F1-score, ROC-AUC, PR-AUC,
and the confusion matrix when evaluating our baseline models.

## Part 4 — Train, Validation, and Test Split

The dataset is divided into training, validation, and test sets to evaluate
the ability of machine learning models to generalize to unseen transactions.

A chronological split is used because credit card fraud detection is a
time-dependent problem. The model learns from historical transactions and
is evaluated on transactions occurring later.

The dataset is divided approximately as follows:

- 70% Training
- 15% Validation
- 15% Test

The training set is used to learn model parameters, the validation set is
used during model development and comparison, and the test set is reserved
for final evaluation.

```python
n = len(X)

train_end = int(0.70 * n)
val_end = int(0.85 * n)

X_train, X_val, X_test = X[:train_end], X[train_end:val_end], X[val_end:]
y_train, y_val, y_test = y[:train_end], y[train_end:val_end], y[val_end:]
```
