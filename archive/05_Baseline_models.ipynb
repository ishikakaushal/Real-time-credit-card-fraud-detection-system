# Real-Time Credit Card Fraud Detection & Risk Intelligence Platform

## Notebook 05 — Baseline Machine Learning Models

### Objective

The objective of this notebook is to establish baseline machine learning models
for credit card fraud detection using the engineered transaction features.

Before developing advanced fraud detection techniques, we first train simple
baseline models to understand how well the engineered features perform.

In this notebook, we will:

- Load the feature-engineered dataset
- Separate features and target variable
- Create training, validation, and test datasets
- Handle the highly imbalanced fraud target appropriately
- Train baseline machine learning models
- Evaluate the models using fraud-detection appropriate metrics
- Compare baseline model performance
- Establish a reference point for future model improvements

### Target Variable

The target variable is:

`is_fraud`

where:

- `0` = legitimate transaction
- `1` = fraudulent transaction

### Evaluation Metrics

Because fraud detection is a highly imbalanced classification problem,
accuracy alone is not sufficient.

We will focus on:

- Precision
- Recall
- F1-score
- ROC-AUC
- PR-AUC
- Confusion Matrix

The baseline results obtained in this notebook will be used as a reference
for subsequent model optimization and fraud-detection improvements.
